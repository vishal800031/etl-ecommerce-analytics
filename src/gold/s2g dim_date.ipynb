```python
date_df = spark.read.table("ecommerce_analytics.silver.sales_orders")
```

```python

from pyspark.sql.functions import *

dim_date = date_df.withColumn(
    "date",
    to_date(col("order_timestamps"))
).select("date").dropDuplicates()
dim_date = dim_date.withColumn(
    "date_key",     
    date_format(col("date"), "yyyyMMdd").cast("int")
).withColumn(
    "day", dayofmonth("date")
).withColumn(
    "month", month("date")
).withColumn(
    "year", year("date")
).withColumn(
    "quarter", quarter("date")
).withColumn(
    "week", weekofyear("date")
)


dim_date.write.mode("overwrite").saveAsTable("ecommerce_analytics.gold.dim_date")
```
