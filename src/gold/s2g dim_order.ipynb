```python
df = spark.read.table("ecommerce_analytics.silver.sales_orders")
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Adding necessary columns
df = df.withColumn("order_timestamps", to_timestamp(col("order_timestamps")))
df = df.withColumn("order_date_key", to_date(col("order_timestamps")))


# Selecting necessary requered columns
dim_order = df.select(
    "order_number",
    "customer_id",
    "order_timestamps",
    "number_of_line_items",
    "order_date_key"
    )
# Saving to final table in gold layer
dim_order.write.mode("overwrite").saveAsTable("ecommerce_analytics.gold.dim_order") 
```
