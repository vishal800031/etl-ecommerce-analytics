```python
dim_products = spark.read.table("ecommerce_analytics.silver.sales")
```

```python
from pyspark.sql.functions import col
dim_final_products = dim_products.select(
    col("id").alias("product_id"),
    "product_name",
    "unit",
    col("product_category").alias("brand_name")
)
dim_final_products = dim_final_products.dropDuplicates(['product_id'])


dim_final_products.write.mode("overwrite").saveAsTable("ecommerce_analytics.gold.dim_products")


```
